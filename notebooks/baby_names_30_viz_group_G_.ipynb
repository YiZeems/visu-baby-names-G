# Brainstorming — 30 visualization ideas for the *Baby Names* project

This notebook presents the broad brainstorming phase of the project. It contains **30 visualization ideas**, organized into three groups:

- **Visualization 1: temporal evolution of baby names**
- **Visualization 2: regional effects**
- **Visualization 3: gender effects**

For each visualization, we explain what it shows, how it represents the data, its strengths, and its limitations.


# Detailed group brainstorming

The goal of the brainstorming session was to generate a wide range of visual design alternatives before selecting a smaller set of candidates for implementation. We intentionally mixed classical visualizations, such as line charts, maps, heatmaps, and rankings, with more original sketches, such as trees, flowers, visual timelines, and icon-based gender representations.

The group discussed several evaluation criteria. The first one was **clarity**: a visualization must be understandable quickly and should answer the project question directly. The second criterion was **relevance**: each design should address one of the three main questions, namely time evolution, regional variation, or gender effects. The third criterion was **comparability**: some visualizations are good for comparing many names, while others are better for telling the story of one selected name. The final criterion was **originality**: we wanted to explore creative designs, but not at the cost of readability.

A major discussion point was the trade-off between **precision** and **communication**. Standard charts are usually more accurate, but they can become visually crowded. More creative designs are often easier to remember and more engaging, but they sometimes need additional explanation. For that reason, the brainstorming phase kept many different directions rather than choosing one solution too early.

For the temporal question, we discussed whether to prioritize exact yearly values or broader name “life cycles.” For the regional question, we debated whether maps were sufficient or whether abstract summaries such as regional profiles could support comparison. For the gender question, we compared simple ratio-based visuals with more analytical views such as scatterplots and heatmaps.

Overall, the brainstorming helped us identify a broad design space before narrowing it down in the second notebook.


# Visualization 1 — Temporal evolution of baby names

This section focuses on how names evolve over time: long-term popularity, sudden trends, short-lived peaks, disappearances, and comebacks.


## 1.1 — Ridgeline popularity waves

Shows how several names evolve over time through stacked popularity curves.

**What it shows:** The x-axis represents time, and each horizontal wave represents one first name. The height of the wave shows how popular the name is during each period.

**How it shows it:** It is very effective for spotting short-lived trends, long-term popularity, and comeback names.

**Strengths:** Very good for spotting peaks, short-lived trends, and stable names.

**Limitations:** Exact values are harder to read, and the chart becomes crowded if too many names are displayed.




## 1.2 — Bump chart of ranks

Shows how the rank of each name changes over time.

**What it shows:** Each line follows one name across years or decades. Moving upward means the name becomes more popular in the ranking.

**How it shows it:** It clearly shows which names rise, fall, enter the top ranks, or disappear.

**Strengths:** Very clear for seeing which names rise, fall, or stay in the top ranks.

**Limitations:** Shows relative rank only, not the actual difference in number of births.




## 1.3 — Interactive time series by name

Shows the detailed evolution of selected names over time.

**What it shows:** The user selects one or several names, and the chart displays their yearly count or percentage of births.

**How it shows it:** It is simple, precise, and easy to understand for a small number of names.

**Strengths:** Simple, precise, and easy to understand for a small number of names.

**Limitations:** Becomes unreadable if too many lines are shown at the same time.




## 1.4 — Name × year heatmap

Shows many names and many years in a single grid.

**What it shows:** Rows represent names, columns represent years, and darker colors indicate higher popularity.

**How it shows it:** It is efficient for detecting periods of strong popularity and comparing many names.

**Strengths:** Efficient for detecting periods of strong popularity and comparing many names.

**Limitations:** Less intuitive than a line chart for following one name precisely.




## 1.5 — Temporal streamgraph

Shows broad popularity waves over time.

**What it shows:** Colored bands change thickness according to the popularity of name groups or categories.

**How it shows it:** It is visually strong and useful for telling a historical trend story.

**Strengths:** Visually strong and useful for telling a broad historical trend story.

**Limitations:** Individual names are less visible because the view focuses on groups.




## 1.6 — Leaf timeline / visual popularity frieze

Shows the life cycle of a name along a timeline.

**What it shows:** Visual marks such as leaves or symbols are placed on the time axis to represent periods of stronger popularity.

**How it shows it:** It is original, memorable, and good for communicating the idea of growth, decline, and comeback.

**Strengths:** Original, memorable, and good for communicating growth, decline, and comeback.

**Limitations:** Less precise than a standard statistical chart.




## 1.7 — Stacked area chart

Shows the evolution of several name groups over time.

**What it shows:** Each colored layer represents a group, and its thickness shows its relative presence in births.

**How it shows it:** It highlights successive periods of dominance between groups of names.

**Strengths:** Highlights successive periods of dominance between groups of names.

**Limitations:** Inner layers can be difficult to compare precisely.




## 1.8 — Name genealogy / tree

Shows names as generations or families evolving over time.

**What it shows:** Branches connect names or name families across periods, suggesting historical succession.

**How it shows it:** It supports a narrative interpretation of name evolution and style replacement.

**Strengths:** Supports a narrative interpretation of name evolution and style replacement.

**Limitations:** Links between names may become subjective if grouping criteria are not clearly defined.




## 1.9 — Small multiple time series

Shows several mini time-series charts, one per name or group.

**What it shows:** Each small chart uses the same time axis, making comparison easier without line overlap.

**How it shows it:** It avoids clutter and keeps each name readable.

**Strengths:** Avoids clutter and keeps each name readable without line overlap.

**Limitations:** Takes more space than a single combined chart.




## 1.10 — Hand-drawn temporal matrix

Summarizes name popularity across time periods.

**What it shows:** Cells in a grid encode popularity using color or mark size.

**How it shows it:** It is compact and useful for comparing many names and periods.

**Strengths:** Compact and useful for comparing many names and periods.

**Limitations:** Can hide fine-grained year-to-year variations.




# Visualization 2 — Regional effects

This section focuses on geographic variation: whether some names are more popular in specific departments or regions, and whether popular names are popular everywhere or only locally.


## 2.1 — Department choropleth map

Shows how popular a name is in each French department.

**What it shows:** Each department is colored according to the name’s birth share or normalized popularity.

**How it shows it:** It is the most intuitive representation for a geographic question.

**Strengths:** The most intuitive representation for a geographic question.

**Limitations:** Raw counts can be misleading because highly populated departments would dominate; normalized values are required.




## 2.2 — Small multiple maps

Compares several names or several years using multiple maps.

**What it shows:** Each map shows one name or one period, ideally with the same color scale.

**How it shows it:** It makes regional patterns directly comparable.

**Strengths:** Makes regional patterns directly comparable across names or periods.

**Limitations:** Requires space and can become dense with too many maps; the color scale must stay identical.




## 2.3 — Small multiple maps

Second example — another name or period with the same color scale.

**Strengths:** Makes regional patterns directly comparable across names or periods.

**Limitations:** Requires space and can become dense with too many maps; the color scale must stay identical.




## 2.4 — Department ranking

Shows which departments are most associated with a given name.

**What it shows:** Departments are ordered by the popularity or percentage of births for the selected name.

**How it shows it:** It is very readable and gives precise top regions.

**Strengths:** Very readable and gives precise top departments or regions.

**Limitations:** Loses the geographic shape of France.




## 2.5 — Tile map / department grid

Shows regional popularity using a simplified grid.

**What it shows:** Each tile represents a department, and its color encodes popularity.

**How it shows it:** It makes department comparison cleaner and reduces visual distortions caused by map area size.

**Strengths:** Makes department comparison cleaner and reduces distortions caused by map area size.

**Limitations:** Real geography is simplified or distorted.




## 2.6 — Annotated regional map with search

Shows a map updated for a searched name.

**What it shows:** The user selects a name, and the map highlights the most relevant regions or departments with annotations.

**How it shows it:** It is a good exploratory interface for regional effects.

**Strengths:** Good exploratory interface for regional effects.

**Limitations:** Depends on interaction and is less powerful as a static image.




## 2.7 — Hand-drawn map for one name

Shows the spatial concentration of one selected name.

**What it shows:** Darker or highlighted areas indicate where the name is more popular.

**How it shows it:** It is simple and easy to understand.

**Strengths:** Simple and easy to understand for one selected name.

**Limitations:** Does not compare several names directly.




## 2.8 — Regional small maps comparison

Compares multiple regional patterns side by side.

**What it shows:** Several maps are shown together to reveal similarities and differences.

**How it shows it:** It is a good compromise between geography and comparison.

**Strengths:** Good compromise between geography and side-by-side comparison.

**Limitations:** The color scale must be consistent to avoid misleading comparisons.




## 2.9 — Regional flower

Summarizes the regional profile of a name.

**What it shows:** Each petal represents a region, and the petal length encodes the name’s popularity in that region.

**How it shows it:** It is original, compact, and easy to compare across names.

**Strengths:** Original, compact, and easy to compare regional profiles across names.

**Limitations:** Loses department-level precision and exact geography.




## 2.10 — Regional podium with map filter

Shows top names in a selected region using a podium-style ranking alongside an interactive map.

**What it shows:** A podium bar chart lists the leading names for the filtered region, while a choropleth map lets the user click a department or region to update the ranking. A related view can highlight the most similar region based on name-distribution similarity.

**How it shows it:** The podium makes regional leaders immediately readable; linking it to the map connects ranking with geography.

**Strengths:** Combines precise ranking with spatial exploration and regional comparison.

**Limitations:** Regional similarity is approximate; the view depends on interaction and is harder to build than a static chart.




# Visualization 3 — Gender effects

This section focuses on differences between girls and boys: whether some names are gender-specific, mixed, or changing over time.


## 3.1 — Girls vs boys scatterplot

Compares name popularity among girls and boys.

**What it shows:** The x-axis represents popularity among girls, while the y-axis represents popularity among boys.

**How it shows it:** It is useful for identifying names that are gender-specific or mixed.

**Strengths:** Useful for identifying names that are gender-specific or mixed.

**Limitations:** Points may overlap, especially for rare names.




## 3.2 — Mixedness index over time

Shows how gender mixedness changes over the years.

**What it shows:** An index measures whether a name is balanced between girls and boys or strongly gendered.

**How it shows it:** It directly addresses the overall question of gender effects.

**Strengths:** Directly addresses the overall question of gender effects over time.

**Limitations:** The mixedness index must be clearly explained to the reader.




## 3.3 — Baby ratio sketch

Shows the girls/boys ratio for a few names in a very intuitive way.

**What it shows:** A baby icon is colored according to the proportion of girls and boys.

**How it shows it:** It is immediately understandable for a non-expert audience.

**Strengths:** Immediately understandable for a non-expert audience.

**Limitations:** Not very precise and works mainly for a small number of examples.




## 3.4 — Gender rings

Shows the gender split for multiple names.

**What it shows:** Each circle is divided into colors representing the girls/boys ratio.

**How it shows it:** It is visually simple and easy to compare for a few names.

**Strengths:** Visually simple and easy to compare for a few names.

**Limitations:** Less suitable for large datasets or exact values.




## 3.5 — Mirrored gender bars

Compares girls and boys for several unisex names.

**What it shows:** Bars extend from a central axis, with one side for girls and the other for boys.

**How it shows it:** It is very readable for comparing gender proportions.

**Strengths:** Very readable for comparing gender proportions across names.

**Limitations:** Does not show how proportions change over time.




## 3.6 — Popularity × mixedness scatterplot

Places names according to both total popularity and gender mixedness.

**What it shows:** The x-axis shows total popularity, and the y-axis shows how balanced the name is between girls and boys.

**How it shows it:** It identifies names that are both popular and truly unisex.

**Strengths:** Identifies names that are both popular and truly unisex.

**Limitations:** The mixedness measure must be explained, and overlapping points may need filtering.




## 3.7 — Gender streams

Shows how girls/boys proportions evolve over time.

**What it shows:** Two colored streams change thickness according to the share of each gender.

**How it shows it:** It is visually effective for showing a gender trend.

**Strengths:** Visually effective for showing a gender trend over time.

**Limitations:** Precise values may be difficult to read.




## 3.8 — Gender trajectory

Shows whether a name becomes more feminine, more masculine, or remains mixed.

**What it shows:** The vertical axis represents years, and the horizontal axis represents the girls/boys ratio.

**How it shows it:** It is clear for following one unisex name over time.

**Strengths:** Clear for following one unisex name over time.

**Limitations:** Works mainly name by name, not for many names at once.




## 3.9 — Heatmap of mixed names

Shows the dominant gender of several names across decades.

**What it shows:** Rows are names, columns are decades, and color indicates girls, boys, or mixed usage.

**How it shows it:** It efficiently compares many unisex names over time.

**Strengths:** Efficient for comparing many unisex names across decades.

**Limitations:** Less detailed than an individual trajectory for one name.




## 3.10 — Gender seesaw over time

Shows how a unisex name tips between girls and boys across key periods.

**What it shows:** A seesaw tilts left toward girls or right toward boys according to the name's gender ratio. Snapshots for different years (e.g. 1900, 1960, 2020) make the shift easy to read at a glance.

**How it shows it:** The tilt angle encodes the girls/boys balance; the user can scrub through time or compare fixed decades side by side.

**Strengths:** Very intuitive and memorable for explaining a gender flip or gradual drift.

**Limitations:** Works best for one name at a time; fixed-year snapshots may hide gradual year-to-year changes.
